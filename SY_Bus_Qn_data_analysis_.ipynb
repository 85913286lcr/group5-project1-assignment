```python
import json
#create a empty list
all_categories = []

for raw in df_all["categories"].dropna():  #go through the categories column one value at a time (after dropping the blank ones), and each time, call the current value raw."
    all_categories.extend(json.loads(raw))

# 2. Build a DataFrame from that list, then keep only unique id/category pairs
category_lookup = pd.DataFrame(all_categories).drop_duplicates().sort_values("id").reset_index(drop=True)

category_lookup
```

```python
before_count = len(df_it)

df_it_filtered = df_it[
    df_it["salary_analysis_ready"]
    .astype(str)
    .str.strip()
    .str.upper()
    .eq("TRUE")
].copy()

after_count = len(df_it_filtered)
reduction = before_count - after_count

print(f"Original records: {before_count}")
print(f"Filtered records: {after_count}")
print(f"Records removed: {reduction}")
```

```python
#To determine the sum of each employmnet type for all industry
summary = df_all.groupby("employmentTypes")["metadata_jobPostId"].count().sort_values(ascending=False)

summary = summary.reset_index()
summary.columns = ["employmentTypes" , "count"]
summary["percentage"] = (summary["count"]/summary["count"].sum()*100).round(1)
summary
```

```python
%pip install plotly
import plotly.express as px

total_postings = summary["count"].sum()

# group the small categories into "Other"
summary["group"] = summary["employmentTypes"].where(summary["percentage"] >= 3, "Other")
chart_data = summary.groupby("group")[["count"]].sum().reset_index()
chart_data["percentage"] = (chart_data["count"] / chart_data["count"].sum() * 100).round(1)
chart_data = chart_data.sort_values("count", ascending=True)

fig = px.bar(
    chart_data,
    x="count", y="group",
    orientation="h",
    text=chart_data["percentage"].astype(str) + "%",
    labels={"count": "Number of postings", "group": "Employment type"},
    title=f"Employment Type Mix — All Industries<br><sup>Total postings: {total_postings:,}</sup>"
)
fig.update_traces(
    marker_color="#4C6EF5",
    textposition="outside",
    cliponaxis=False
)

fig.update_layout(
    xaxis=dict(
        title=dict(
            text="Number of postings",
            standoff=60
        ),
        range=[0, chart_data["count"].max() * 1.15]
    ),
    yaxis_title=None,
    plot_bgcolor="white",
    margin=dict(b=140, r=80)
)


fig.add_annotation(
    text="Other includes: Part Time, Temporary, Internship/Attachment, Freelance, Flexi-work",
    xref="paper", yref="paper",
    x=0, y=-0.2,
    showarrow=False,
    font=dict(size=11, color="gray"),
    align="left"
)

fig.show()
```

```python
summary_it = df_it_filtered["employmentTypes"].value_counts().reset_index()
summary_it.columns = ["employmentTypes", "count"]
summary_it["percentage"] = (summary_it["count"] / summary_it["count"].sum() * 100).round(1)

total_it_postings = summary_it["count"].sum()

summary_it["group"] = summary_it["employmentTypes"].where(summary_it["percentage"] >= 3, "Other")
chart_data_it = summary_it.groupby("group")[["count"]].sum().reset_index()
chart_data_it["percentage"] = (chart_data_it["count"] / chart_data_it["count"].sum() * 100).round(1)
chart_data_it = chart_data_it.sort_values("count", ascending=True)

fig_it = px.bar(
    chart_data_it,
    x="count", y="group",
    orientation="h",
    text=chart_data_it["percentage"].astype(str) + "%",
    labels={"count": "Number of postings", "group": "Employment type"},
    title=f"Employment Type Mix — IT Postings<br><sup>Total postings: {total_it_postings:,}</sup>"
)
fig_it.update_traces(
    marker_color="#12877F",
    textposition="outside",
    cliponaxis=False
)

fig_it.update_layout(
    xaxis=dict(
        title=dict(
            text="Number of postings",
            standoff=60
        ),
        range=[0, chart_data_it["count"].max() * 1.15]
    ),
    yaxis_title=None,
    plot_bgcolor="white",
    paper_bgcolor="white",
    margin=dict(b=140, r=80)
)


fig_it.add_annotation(
    text="Other includes:<br>"
         "Internship/Attachment, Temporary, Part Time, "
         "Freelance, Flexi-work",
    xref="paper",
    yref="paper",
    x=0,
    y=-0.34,
    showarrow=False,
    font=dict(size=11, color="gray"),
    align="left"
)

fig_it.show()
```

```python
def summarize(data):
    s = data["employmentTypes"].value_counts().reset_index()
    s.columns = ["employmentTypes", "count"]
    s["percentage"] = (s["count"] / s["count"].sum() * 100).round(1)
    s["group"] = s["employmentTypes"].where(s["percentage"] >= 3, "Other")
    g = s.groupby("group")[["count"]].sum().reset_index()
    g["percentage"] = (g["count"] / g["count"].sum() * 100).round(1)
    return g


all_summary = summarize(df_all).assign(dataset="All Industries")
it_summary = summarize(df_it_filtered).assign(dataset="IT")

combined = pd.concat([all_summary, it_summary])

# fix the bar order so it reads Permanent -> Full Time -> Contract -> Other
order = ["Permanent", "Full Time", "Contract", "Other"]
combined["group"] = pd.Categorical(combined["group"], categories=order, ordered=True)
combined = combined.sort_values("group")

fig = px.bar(
    combined,
    x="percentage",
    y="group",
    color="dataset",
    barmode="group",
    orientation="h",
    text=combined["percentage"].astype(str) + "%",
    labels={
        "percentage": "% of postings",
        "group": "Employment type",
        "dataset": ""
    },
    color_discrete_map={
        "All Industries": "#4C6EF5",  # blue
        "IT": "#12877F"                # green
    },
    title=f"Employment Type Mix: All Industries "
          f"(n={len(df_all):,}) vs IT "
          f"(n={len(df_it_filtered):,})"
)

fig.update_traces(
    textposition="outside",
    cliponaxis=False
)

fig.update_layout(
    xaxis_title=None,
    yaxis_title=None,
    plot_bgcolor="white",
    margin=dict(b=90)
)
fig.add_annotation(
    text="Note: figures are % share within each dataset.<br>Other includes: Part Time, Temporary, Internship/Attachment, Freelance, Flexi-work",
    xref="paper", yref="paper",
    x=0, y=-0.2,
    showarrow=False,
    font=dict(size=11, color="gray"),
    align="left"
)
fig.show()
```
